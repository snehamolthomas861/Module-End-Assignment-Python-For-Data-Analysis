 Cleaning Missing Data

```python
# Checking median values
print("Age median:", df['age'].median())
print("Likes median:", df['likes'].median())
print("Comments median:", df['comments'].median())
print("Shares median:", df['shares'].median())
```

Handling Missing Values

```python
# Numerical columns - fill missing values with median
df['age'] = df['age'].fillna(df['age'].median())
df['likes'] = df['likes'].fillna(df['likes'].median())
df['comments'] = df['comments'].fillna(df['comments'].median())
df['shares'] = df['shares'].fillna(df['shares'].median())

# Categorical columns - fill missing values with mode
df['gender'] = df['gender'].fillna(df['gender'].mode()[0])
df['sentiment'] = df['sentiment'].fillna(df['sentiment'].mode()[0])
```

Final Insights

Content Performance

Which post types have highest engagement?


```python
post_type_engagement = df.groupby('post_type')['engagement_score'].mean().sort_values(ascending=False)

print(post_type_engagement)
```

Best-performing content category?

```python
category_engagement = (
    df.groupby('post_category')['engagement_score']
      .mean()
      .sort_values(ascending=False)
)

print(category_engagement)
```

Which countries have highest avg engagement rate?

```python
country_engagement = (
    df.groupby('country')['engagement_rate']
      .mean()
      .sort_values(ascending=False)
)

print(country_engagement)
```

User Trends

How age affects engagement

```python
df['age_group'] = pd.cut(
    df['age'],
    bins=[0, 18, 25, 35, 45,55,100],
    labels=['Under 18', '18-25', '26-35', '36-45', '46-55','56+']
)
```

```python
age_engagement = df.groupby('age_group', observed=True)['engagement_score'].mean()

print(age_engagement)
```

Performance difference for verified accounts

```python
verified_performance = (
    df.groupby('is_verified')['engagement_score']
      .mean()
)

print(verified_performance)
```

Final Insights:

User Trends

Engagement varies across different age groups. The Under 18 group has the highest average engagement, while 56+ group has the lowest. This suggests that engagement levels differ across age groups in the dataset.

The analysis compares the engagement performance of verified and non-verified accounts.Non-verified accounts have a higher average engagement score when compared with verified accounts.

Content Performance:

Among the different post types, Video posts have the highest average engagement score. This indicates that this type of content generates more combined likes, comments, and shares compared with the other post types in the dataset.

Music is the best-performing content category. This indicates that posts in this category receive higher combined likes, comments, and shares compared with other content categories in the dataset.

Brazil has the highest average engagement rate , indicating that posts from this country achieve a higher level of engagement relative to their impressions compared with other countries in the dataset.

Behavioral Insights:

Best time of day for impressions

```python
df['post_hour'] = df['posted_at'].dt.hour
```

```python
hourly_impressions = (
    df.groupby('post_hour')['impression_count']
      .mean()
      .sort_values(ascending=False)
)

print(hourly_impressions)
```

Insight: In this dataset, posts made at 12 AM (midnight) have an average of 50,013.73 impressions, making as the highest-impression hour.

Device type impact on watch time

```python
device_watch_time = (
    df.groupby('device_type')['watch_time_sec']
      .mean()
      .sort_values(ascending=False)
)

print(device_watch_time)
```

The analysis shows that Mobile has the highest average watch time of approximately 4087.83 seconds. This indicates that users on this device type spend more time watching content on average compared with the other device types in the dataset.

Sentiment Analysis:

Which sentiment performs best

```python
sentiment_engagement = (
    df.groupby('sentiment')['engagement_score']
      .mean()
      .sort_values(ascending=False)
)

print(sentiment_engagement)
```

Negative posts have the highest average engagement. This indicates that posts with this sentiment receive higher engagement, based on the combined likes, comments, and shares in the dataset.



Behavior of negative/neutral sentiment posts


```python
sentiment_behavior = df.groupby('sentiment').agg({
    'likes': 'mean',
    'comments': 'mean',
    'shares': 'mean',
    'engagement_score': 'mean',
    'impression_count': 'mean'
})

print(sentiment_behavior)
```

Negative,Positive and neutral sentiment posts show different engagement patterns.Negative posts have a higher average engagement score, compared with other sentiment. This suggests that sentiment is associated with differences in engagement levels in this dataset.
